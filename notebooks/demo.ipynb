# FinAgent — quick walkthrough

A multi-agent financial-research system (LangGraph + Claude). This notebook shows a
few calls. It runs **offline with the deterministic mock provider** (no API key); set
`LLM_PROVIDER=anthropic` in your `.env` for real Claude runs.


````python
import os
os.environ.setdefault('LLM_PROVIDER','mock')       # offline; use 'anthropic' for real
os.environ.setdefault('SEC_EDGAR_MODE','cache')     # serve SEC from committed fixtures
os.environ.setdefault('WEB_SEARCH_PROVIDER','stub')
from finagent.graph import run_query
````

## 1. A single-tool question


````python
s = run_query("What was Apple's revenue in FY2025?")
print('tools:', s.tools_used)
print('answer:', s.final_answer.answer)
print('key_numbers:', s.final_answer.key_numbers)
print('confidence:', s.final_answer.confidence)
````

## 2. A comparison — watch the reflection loop fire

Two companies: the planner fetches one, `reflect` notices the other is missing and
loops back. `reflection_notes` is the evidence.


````python
s = run_query("Compare Apple's and Microsoft's net margin.")
for tc in s.tool_calls:
    print('tool:', tc.tool, tc.input)
print('iterations looped:', s.iteration)
print('reflection_notes:', s.reflection_notes)
````
